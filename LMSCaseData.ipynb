```python
import numpy as np
import matplotlib.pyplot as plt
import random

# 定义nodes（假设节点0是中心节点，其他节点是1-TN-1）
TN =  12+1
unique_nodes = [i for i in range(TN)]

# The demand range can be customized according to survey or other datasource, we randomly set a simple range

def demand_data(seed=None): # random seed
    # Define ranges for each phase
    range1 = [  # 中等水平范围 (3-7)
        [3, 6], [4, 7], [3, 6], [4, 7], [3, 6],
        [4, 7], [3, 6], [4, 7], [3, 6], [4, 7],
        [3, 6], [4, 7]
    ]
    range2 = [  # 高峰水平范围 (6-12)
        [7, 11], [8, 12], [7, 11], [8, 12], [7, 11],
        [8, 12], [7, 11], [8, 12], [7, 11], [8, 12],
        [7, 11], [8, 12]
    ]
    range3 = [  # 低峰水平范围 (0-3)
        [1, 3], [0, 3], [1, 3], [0, 3], [1, 3],
        [0, 3], [1, 3], [0, 3], [1, 3], [0, 3],
        [1, 3], [0, 3]
    ]

    # Initialize demand dictionary
    demand = {}
    if seed is not None:
        random.seed(seed)

    # Generate demand data for each node
    for i, node in enumerate(unique_nodes[1:]):  # Exclude the center node (node 0)
        # Get ranges for the current node
        phase_1_range = range1[i]
        phase_2_range = range2[i]
        phase_3_range = range3[i]

        # Generate demands for each phase
        phase_1_demand = [random.randint(*phase_1_range) for _ in range(5)]
        phase_2_demand = [random.randint(*phase_2_range) for _ in range(10)]
        phase_3_demand = [random.randint(*phase_3_range) for _ in range(5)]

        # Combine all phases
        demand[node] = phase_1_demand + phase_2_demand + phase_3_demand

    return demand

    print(demand_data(seed=42))
```
